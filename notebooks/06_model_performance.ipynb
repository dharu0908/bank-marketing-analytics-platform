```python
from pyspark.sql import Row
from pyspark.sql.functions import current_timestamp, round as spark_round

performance_data = [
    Row(
        model_name="Logistic Regression",
        auc_roc=0.8717,
        precision=0.8946,
        recall=0.8243,
        f1_score=0.8473,
        target_auc=0.75,
        met_target="Yes"
    ),
    Row(
        model_name="Random Forest",
        auc_roc=0.8863,
        precision=0.8975,
        recall=0.8164,
        f1_score=0.8421,
        target_auc=0.75,
        met_target="Yes"
    )
]

model_performance_df = spark.createDataFrame(performance_data)

model_performance_df = (
    model_performance_df
    .withColumn("auc_roc", spark_round("auc_roc", 4))
    .withColumn("precision", spark_round("precision", 4))
    .withColumn("recall", spark_round("recall", 4))
    .withColumn("f1_score", spark_round("f1_score", 4))
    .withColumn("training_timestamp", current_timestamp())
)

display(model_performance_df)

model_performance_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("bank_model_performance")

print("Created Delta table: bank_model_performance")
```
